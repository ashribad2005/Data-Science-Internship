# Task 1: Data Visualization using Titanic Dataset

## Objective

#The objective of this task is to explore the Titanic dataset using various data visualization techniques. 
#Different charts are used to understand the distribution of passengers, relationships between variables, 
#and important patterns in the data.


```python

# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside notebook
%matplotlib inline

# Improve plot appearance
sns.set_style("whitegrid")

# Load Titanic dataset

df = pd.read_csv("../datasets/titanic.csv")

# Display first five rows
df.head()
print("Shape of Dataset:", df.shape)
print("\nColumn Names:")
print(df.columns)
print("\nDataset Information:")
df.info()
print("\nStatistical Summary:")
display(df.describe())
df.isnull().sum()
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=30, kde=True)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

### Observation
#Most passengers were between 20 and 40 years of age.
#Very few passengers were above 70 years.
#The age distribution is slightly right-skewed.

plt.figure(figsize=(6,4))
sns.countplot(x="Sex", data=df)
plt.title("Gender Distribution")
plt.show()

### Observation
#Most passengers were male. The number of male passengers was significantly higher than female passengers.

plt.figure(figsize=(6,4))
sns.countplot(x="Pclass", data=df)
plt.title("Passenger Class Distribution")
plt.show()

### Observation
#Most passengers travelled in Third Class.

plt.figure(figsize=(6,4))
sns.countplot(x="Survived", data=df)
plt.title("Survival Distribution")
plt.xticks([0,1],["Not Survived","Survived"])
plt.show()

plt.figure(figsize=(7,5))
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by Gender")
plt.legend(["No","Yes"])
plt.show()

### Observation
#Female passengers had a much higher survival rate compared to male passengers.

plt.figure(figsize=(8,5))
sns.histplot(df["Fare"], bins=40)
plt.title("Fare Distribution")
plt.show()

plt.figure(figsize=(7,5))
sns.boxplot(y=df["Age"])
plt.title("Boxplot of Passenger Age")
plt.show()

### Observation
#The boxplot indicates the presence of some age outliers, especially among older passengers.

plt.figure(figsize=(8,6))
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

# Conclusion

#From the visual analysis of the Titanic dataset:
#Most passengers were between 20 and 40 years old.
#Male passengers outnumbered female passengers.
#Most passengers travelled in Third Class.
#Female passengers had a significantly higher survival rate.
#Fare values were highly skewed, with a few passengers paying much higher fares.
